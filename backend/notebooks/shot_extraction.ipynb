```python
import json
from datetime import datetime
from dataclasses import asdict, is_dataclass

from src.services.processing.shot_extraction.service import ShotExtractionService
```

```python
shot_extraction_service = ShotExtractionService()
shot_extraction_service.get_available_providers()
```

```python
shot_extraction_service.register_extractor(
    extractor_name="autoshot",
    provider_name="autoshot"
)
```

```python
shot_result = shot_extraction_service.extract_shots_from_video(
    video_path="sample/video/sample_video.mp4",
    extractor_name="autoshot"
)
```

```python
def default_serializer(obj):
    if isinstance(obj, datetime):
        return obj.isoformat()
    if is_dataclass(obj):
        return asdict(obj)
    raise TypeError(f"Type {type(obj)} not serializable")

with open("sample/processed/shots/sample_video.json", "w", encoding="utf-8") as f:
    json.dump(asdict(shot_result), f, default=default_serializer, ensure_ascii=False, indent=4)
```
